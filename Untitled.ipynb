```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

# Constantes de notre modèle
F = 85.0              # Coûts fixes mensuels (USD)
C_tour = 0.01283      # Coût variable moyen par tour (API + Compute)
jours_par_mois = 30

# Définition des plages de valeurs
n = np.linspace(100, 2000, 50)  # De 100 à 2000 utilisateurs
T = np.linspace(2, 50, 50)      # De 2 à 12 tours par jour
n_grid, T_grid = np.meshgrid(n, T)

# Fonction de coût : C = F + (n * T * 30 * C_tour)
cost_grid = F + (n_grid * T_grid * jours_par_mois * C_tour)

# Création du graphique 3D
fig = plt.figure(figsize=(12, 8))
ax = fig.add_subplot(111, projection='3d')

# Tracer la surface
surf = ax.plot_surface(n_grid, T_grid, cost_grid, cmap='viridis', edgecolor='none', alpha=0.9)

# Labels
ax.set_xlabel('Nombre d\'utilisateurs (n)')
ax.set_ylabel('Tours par jour (T)')
ax.set_zlabel('Coût Mensuel (USD)')
ax.set_title('Surface de Coût : AgriConnect (Groq + Twilio)')

# Ajout d'une barre de couleur
fig.colorbar(surf, shrink=0.5, aspect=5, label='Coût en USD')

plt.tight_layout()
plt.show()
```
