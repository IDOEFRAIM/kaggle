# EDA - House Price Prediction
### This notebook presents a step-by-step exploratory data analysis (EDA) for the house price prediction dataset.

## Workflow:

### 1. Data loading and preview
### 2. Target variable analysis
### 3. Missing values and duplicates
### 4. Feature summary (categorical/numerical)
### 5. Distribution and correlation analysis
### 6. Visualization of key features

```python
# 1. Data loading and preview
import os
import numpy as np
import pandas as pd 
import seaborn as sns
import matplotlib.pyplot as plt

TRAIN_PATH = 'dataset/train.csv'
TEST_PATH = 'dataset/test.csv'

TARGET = 'SalePrice'
ID_COL = 'Id'

assert os.path.exists(TRAIN_PATH), f"The path:{TRAIN_PATH} doesnt exist. Check in"
assert os.path.exists(TEST_PATH), f"The path:{TEST_PATH} doesnt exist. Check in"

train = pd.read_csv(TRAIN_PATH)
test = pd.read_csv(TEST_PATH)

print(f"The train shape: {train.shape}")
print(f"The test shape: {test.shape}")

display(train.head())
```

# Check for nul value and na

```python
# 2. Target variable analysis
print(train[TARGET].describe())
plt.figure(figsize=(8,5))
sns.histplot(train[TARGET], kde=True)
plt.title('Distribution of SalePrice')
plt.show()
```

```python
# 3. Missing values and duplicates
print('Missing values per column:')
print(train.isnull().sum().sort_values(ascending=False))
print('\nDuplicate rows:', train.duplicated().sum())
```

```python
train.isna().sum()
```

```python
train.duplicated().sum()
```

```python
# 4. Feature summary (categorical/numerical)
num_features = train.select_dtypes(include=[np.number]).columns.tolist()
cat_features = train.select_dtypes(exclude=[np.number]).columns.tolist()
print(f'Numerical features ({len(num_features)}): {num_features}')
print(f'Categorical features ({len(cat_features)}): {cat_features}')
print('\nNumerical summary:')
display(train[num_features].describe())
print('\nCategorical summary:')
for col in cat_features:
    print(f'{col}: {train[col].nunique()} unique values')
```

```python
FEATURES_COLS = train.columns
DROP_COLS = [ID_COL,TARGET]
USED_COLS = [c for c in FEATURES_COLS if c not in DROP_COLS]
```

```python
# 5. Correlation analysis and target distribution
corr = train.corr()
plt.figure(figsize=(12,8))
sns.heatmap(corr, cmap='coolwarm', center=0)
plt.title('Correlation Heatmap')
plt.show()

plt.figure(figsize=(8,5))
sns.boxplot(x=train[TARGET])
plt.title('SalePrice Boxplot')
plt.show()
```
